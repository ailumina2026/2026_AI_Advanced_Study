**Classification vs Detection의 차이:**
- Classification: 이미지 전체를 하나의 클래스로 분류 ("이 이미지는 무엇?")
- Detection: 이미지 내 여러 객체의 위치와 클래스를 탐지 ("어디에 무엇이 있는가?")

**Detection 학습의 특징:**
- `config.yaml` 파일이 필요합니다 (Classification은 폴더 구조만 사용)
- 결과는 `runs/detect/train/`에 저장됩니다

#### 4️⃣ 하이퍼파라미터 설정 🎯

AI 모델의 성능을 결정하는 중요한 설정값들을 지정합니다.

- **MODEL_SIZE**: 모델의 크기 (Detection 모델은 `-cls` 접미사가 없습니다!)
  - `'yolo11n.pt'`: Nano (가장 빠름, 정확도 낮음)
  - `'yolo11s.pt'`: Small (빠름, 정확도 보통)
  - `'yolo11m.pt'`: Medium (보통, 정확도 높음)
  - `'yolo11l.pt'`: Large (느림, 정확도 매우 높음)

- **EPOCHS**: 학습 횟수
- **IMG_SIZE**: 이미지 크기 (Detection은 보통 640 사용)

````python
# Detection 모델 선택 (Classification과 달리 -cls가 없음!)
MODEL_SIZE = 'yolo11n.pt'

# 학습 epoch 수
EPOCHS = 10  # 실험해보기: 5, 30

# 이미지 크기 (Detection은 640이 표준)
IMG_SIZE = 640  # 실험해보기: 320, 480

# 배치 크기
BATCH_SIZE = 16  # 실험해보기: 8, 32

print("📋 현재 설정:")
print(f"  - 모델 크기: {MODEL_SIZE}")
print(f"  - Epochs: {EPOCHS}")
print(f"  - 이미지 크기: {IMG_SIZE}")
print(f"  - 배치 크기: {BATCH_SIZE}")
````

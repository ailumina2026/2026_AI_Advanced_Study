#### 3️⃣ 데이터셋 경로 및 규모 확인

모델을 학습시키기 전, **준비된 데이터가 총 몇 장인지** 확인하는 과정은 필수입니다.

YOLO Classification은 Detection과 달리 **config.yaml 파일이 필요 없습니다.**  
폴더 구조 자체가 클래스 정보를 정의합니다:
```
data/
├── train/
│   ├── Tomato_healthy/
│   ├── Tomato_Bacterial_spot/
│   └── ...
├── val/
└── test/
```

````python
# 🎯 [미션] 데이터 폴더의 경로를 설정하세요.
# 힌트: 상위 폴더의 data 디렉토리를 가리킵니다.
DATA_PATH = _______

print(f"📘 데이터 경로: {DATA_PATH}")
print("-" * 50)

# 각 데이터셋(학습/검증/테스트) 별 이미지 개수 세기
for split in ['train', 'val', 'test']:
    split_path = os.path.join(DATA_PATH, split)
    if os.path.exists(split_path):
        classes = [d for d in os.listdir(split_path) if os.path.isdir(os.path.join(split_path, d))]
        total_images = 0
        print(f"\n📂 {split.upper()} 데이터:")
        for cls in sorted(classes):
            cls_path = os.path.join(split_path, cls)
            image_count = len(glob.glob(os.path.join(cls_path, '*.[jJ][pP][gG]'))) + \
                          len(glob.glob(os.path.join(cls_path, '*.[pP][nN][gG]')))
            total_images += image_count
            print(f"   - {cls}: {image_count}장")
        print(f"   총: {total_images}장")

print("\n" + "-" * 50)
````

#### 4️⃣ 하이퍼파라미터 설정 🎯

AI 모델의 성능을 결정하는 중요한 설정값들을 지정합니다. 여러분이 원하는 값으로 변경해 볼 수 있습니다.

- **MODEL_SIZE**: 모델의 크기와 태스크 유형
  - `'yolo11n-cls.pt'`: Nano Classification (가장 빠름, 정확도 낮음)
  - `'yolo11s-cls.pt'`: Small Classification (빠름, 정확도 보통)
  - `'yolo11m-cls.pt'`: Medium Classification (보통, 정확도 높음)
  - `'yolo11l-cls.pt'`: Large Classification (느림, 정확도 매우 높음)

- **EPOCHS**: 학습 횟수 (너무 적으면 학습이 안 되고, 너무 많으면 시간이 오래 걸림)

- **IMG_SIZE**: 이미지 크기 (Classification은 보통 224 사용)

````python
# 🎯 [미션] 분류(Classification) 모델을 사용하려면 어떤 접미사를 붙여야 할까요?
MODEL_SIZE = 'yolo11n-_____.pt'

# 학습 epoch 수
EPOCHS = 10  # 실험해보기: 5, 30

# 이미지 크기 (Classification은 224가 표준)
IMG_SIZE = 224  # 실험해보기: 128, 256

# 배치 크기
BATCH_SIZE = 32  # 실험해보기: 16, 64

print("📋 현재 설정:")
print(f"  - 모델 크기: {MODEL_SIZE}")
print(f"  - Epochs: {EPOCHS}")
print(f"  - 이미지 크기: {IMG_SIZE}")
print(f"  - 배치 크기: {BATCH_SIZE}")
````
